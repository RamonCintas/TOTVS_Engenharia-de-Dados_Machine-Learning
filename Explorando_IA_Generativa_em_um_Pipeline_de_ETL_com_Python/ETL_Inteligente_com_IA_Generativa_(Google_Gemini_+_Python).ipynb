# 🚀 ETL Inteligente com IA Generativa (Google Gemini + Python)

Projeto prático demonstrando a aplicação de IA Generativa em um pipeline completo de ETL,
utilizando Python e Google Gemini para criação de mensagens personalizadas.

## 📌 Arquitetura do Pipeline

**Extract**
- Leitura de dados simulados localmente
- Estruturação padronizada dos usuários

**Transform**
- Geração de mensagens personalizadas com Google Gemini
- Engenharia de Prompt com restrições (máx. 100 caracteres)
- Personalização baseada em perfil financeiro
- Tratamento e validação automática de saída

**Load**
- Persistência local em JSON

```python
from google.colab import userdata
from google import genai

google_api_key = userdata.get("API-GEMINI-KEY")

if not google_api_key:
    raise ValueError("API key não encontrada nos Secrets do Colab.")

client = genai.Client(api_key=google_api_key)

print("API configurada com sucesso!")

for model in client.models.list():
    print(model.name)
```

## 🟦 EXTRACT
Simulando dados fictícios localmente

```python
users = [
        {"id": 1, "name": "Ana Silva", "account": {"balance": 1500.75}, "news": []},
        {"id": 2, "name": "Carlos Souza", "account": {"balance": 8200.00}, "news": []},
        {"id": 3, "name": "Mariana Costa", "account": {"balance": 350.40}, "news": []},
    ]
print("Usuários simulados carregados!")
```

## 🟨 TRANSFORM (Gemini)

```python
def generate_ai_news(user):

    balance = user.get("account", {}).get("balance", 0)

    if balance >= 5000:
        perfil = "cliente com alto potencial de investimento"
    elif balance >= 1000:
        perfil = "cliente com potencial de crescimento financeiro"
    else:
        perfil = "cliente iniciando sua jornada financeira"

    prompt = f"""
    Gere apenas uma mensagem de marketing bancário.
    Máximo 100 caracteres.
    Sem explicações.
    Sem múltiplas opções.
    Sem aspas.
    Sem quebra de linha.
    Retorne somente o texto final.

    Cliente: {user['name']}
    Perfil: {perfil}
    """

    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt
    )

    mensagem = response.text.strip().replace("\n", " ")
    return mensagem[:100]

for user in users:
    try:
        user["news"] = []

        news = generate_ai_news(user)

        user["news"].append({
            "icon": "💰",
            "description": news
        })

        print(f"{user['name']}: {news}")

    except Exception as e:
        print(f"Erro ao gerar mensagem para {user['name']}: {e}")
```

## 🟩 LOAD
Salvando resultado em JSON

```python
import json
import os

file_name = "Usuarios_Atualizados.json"

with open(file_name, "w", encoding="utf-8") as f:
    json.dump(users, f, indent=2, ensure_ascii=False)

full_path = os.path.abspath(file_name)

print("Arquivo salvo com sucesso!")
print("Caminho completo:", full_path)

print("Diretório atual:", os.getcwd())
```
